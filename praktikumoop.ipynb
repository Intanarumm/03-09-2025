```python
class Kucing:
    # Constructor
    def __init__(self, nama, warna):
        self.nama = nama   # Attribute
        self.warna = warna # Attribute

    # Method
    def suara(self):
        return f"{self.nama} berkata: Meow!"

# Membuat Object (Instansiasi)
kucing1 = Kucing("Luna", "Putih")
kucing2 = Kucing("Garfield", "Oranye")
# Mengakses Attribute dan Method
print(kucing1.nama) # Output: Luna
print(kucing1.warna) # Output: Putih
print(kucing1.suara()) # Output: Garfield berkata: Meow!
print(kucing2.nama) # Output: Garfield
print(kucing2.warna) # Output: Oranye
print(kucing2.suara()) # Output: Garfield berkata: Meow!
```

```python
class AkunBank:
    def __init__(self, saldo):        # <-- pakai __ (dua underscore) di kedua sisi
        self.__saldo = saldo  # Private attribute

    def cek_saldo(self):
        return self.__saldo

    def setor(self, jumlah):
        if jumlah > 0:
            self.__saldo += jumlah

    def tarik(self, jumlah):
        if jumlah > 0 and jumlah <= self.__saldo:
            self.__saldo -= jumlah
        else:
            print("Jumlah yang dimasukkan tidak valid atau saldo tidak mencukupi.")

rekening = AkunBank(1000)
rekening.setor(500)
rekening.tarik(200)
print(rekening.cek_saldo()) # Output: 1500
# print(rekening.__saldo)
```

```python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name = name  # Nama Hero
        self.hp = hp  # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# -- Main Program --
# Membuat Object (Instansiasi)
hero1 = Hero("Layla", 500, 15)
hero2 = Hero("Zilong", 120, 20)

# Memanggil Method
hero1.info()
hero2.info()

# Tambah kode Output di akhir program
print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2)  # Layla menyerang Zilong
hero2.serang(hero1)  # Zilong membalas
```

Analisis
Kalau parameter lawan cuma berupa nama doang (string "Zilong"), ya tampilannya cuman nama doang/bukan objek, jadi dengan nama doang itu si "Zilong" gabisa ngapa ngapain soalnya bukan objek

```python
# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)
eudora.info()
eudora.serang(balmond)          # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond)  # Skill khusus Mage
```

```python
from abc import ABC, abstractmethod

class Hero(ABC):
    def __init__(self, nama, hp, power):
        self.nama = nama
        self.__hp = hp
        self.power = power

    def get_hp(self):
        return self.__hp

    def diserang(self, damage):
        self.__hp = max(0, self.__hp - damage)
        print(f"{self.nama} kena damage {damage}. Sisa HP: {self.get_hp()}")

    @abstractmethod
    def serang(self, lawan):
        pass

    def info(self):
        print(f"{self.nama} | HP: {self.get_hp()} | Power: {self.power}")


class Fighter(Hero):
    def serang(self, lawan):
        print(f"{self.nama} (Fighter) menebas {lawan.nama}!")
        lawan.diserang(self.power)


class Mage(Hero):
    def __init__(self, nama, hp, power, mana):
        super().__init__(nama, hp, power)
        self.mana = mana

    def serang(self, lawan):
        if self.mana >= 20:
            self.mana -= 20
            print(f"{self.nama} (Mage) melempar Fireball ke {lawan.nama}!")
            lawan.diserang(self.power * 2)
        else:
            print(f"{self.nama} mana habis, serangan biasa saja.")
            lawan.diserang(self.power // 2)

hero1 = Fighter("Zilong", 120, 20)
hero2 = Mage("Eudora", 80, 15, 100)

hero1.info()
hero2.info()

print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2)
hero2.serang(hero1)
```
